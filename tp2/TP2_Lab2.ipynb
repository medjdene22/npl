## Exercise 1 — Characters and a custom cleaner

```python
from collections import Counter

text = "NLP is fun!"
characters = list(text)

print("1) Characters:", characters)
print("   Number of characters:", len(characters))


no_spaces = [char for char in characters if char != " "]

print("\n2) Without spaces:", no_spaces)
print("   Number of characters:", len(no_spaces))


def char_frequency(text):
    freq = {}

    for char in text.lower():
        if char.isalpha():
            freq[char] = freq.get(char, 0) + 1

    return freq


sentence = "Natural language processing is fascinating!"
freq = char_frequency(sentence)

print("\n3) freq:", freq)

top_three = sorted(freq.items(), key=lambda item: item[1], reverse=True)[:3]

print("\n4) Three most frequent letters:")
for letter, count in top_three:
    print(f"   {letter!r}: {count}")


def char_frequency2(text):
    return Counter(char for char in text.lower() if char.isalpha())


top_three = char_frequency2(sentence).most_common(3)

print("\n5) Three most frequent letters (Counter):")
for letter, count in top_three:
    print(f"   {letter!r}: {count}")
```

## Exercise 2 — Word tokenization with NLTK

```python
from nltk.tokenize import word_tokenize

text = "Natural language processing is interesting."
tokens = word_tokenize(text)

print("1) tokens:", tokens, "lenth:", len(tokens))


def naive_tokenize(text):
    return text.split()

naive_tokens = naive_tokenize(text)

print("2) naive tokens:", naive_tokens, "lenth:", len(naive_tokens))


hard_text = "Well, that's... surprising (and a bit odd)!"
naive_hard = naive_tokenize(hard_text)
nltk_hard = word_tokenize(hard_text)

print("3) Harder sentence:", hard_text)
print("   naive tokens:", naive_hard)
print("   nltk tokens:", nltk_hard)

only_naive = [token for token in naive_hard if token not in nltk_hard]
only_nltk = [token for token in nltk_hard if token not in naive_hard]

print("   Tokens only in naive output: ", only_naive)
print("   Tokens only in NLTK output:  ", only_nltk)


print("4) Comparison table:")

punct = [".", ",", "!", "?", "(", ")", "..."]
sentences = [text, hard_text]

for sentence in sentences:
    for name, tokens in [("naive", naive_tokenize(sentence)), ("nltk", word_tokenize(sentence))]:
        print(name, "|", len(tokens), "|", "yes" if any(token in punct for token in tokens) else "no")
```

## Exercise 3 — Word-tokenization difficulties

```python
from nltk.tokenize import TreebankWordTokenizer, WordPunctTokenizer, word_tokenize

# 1. word_tokenize()
text = "I don't know. It's useful, isn't it?"
print("1) word_tokenize(text):", word_tokenize(text))

# 2+3. TreebankWordTokenizer and WordPunctTokenizer
expressions = ["don't", "isn't", "New York", "state-of-the-art", "twenty-one"]
tb = TreebankWordTokenizer()
wp = WordPunctTokenizer()

print("\n2+3) TreebankWordTokenizer and WordPunctTokenizer:")
for e in expressions:
    print(e, "|", word_tokenize(e), "|", tb.tokenize(e), "|", wp.tokenize(e))
```

## Mini-challenge — Comparing tokenizers on noisy text

```python
from nltk.tokenize import TweetTokenizer, word_tokenize

text = "omgggg this lab is SO gooood!! can't wait for lab3 :) check https://nlp.univ.dz #NLP @prof_ahmed"

ws    = text.split()
wt    = word_tokenize(text)
tweet = TweetTokenizer().tokenize(text)

for name, toks in [("split", ws), ("word_tokenize", wt), ("TweetTokenizer", tweet)]:
    print(name, len(toks))
    print(toks, "\n")

print("comparison table:")
print(f"{'Tokenizer':<18}{'Tokens':<8}{'Hashtag':<9}{'URL':<6}{'Emoticon':<9}")
print("-" * 50)

for name, toks in [("split", ws), ("word_tokenize", wt), ("TweetTokenizer", tweet)]:
    print(f"{name:<18}{len(toks):<8}{'no' if '#' in toks else 'yes':<9}{'no' if 'https' in toks else 'yes':<6}{'yes' if ':)' in toks else 'no':<9}")
```
